```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
DRN_cell_11_char = pd.read_csv('DRN-serotonin_Female_spikes\CELL 11 DEEP 4614_Spikes.csv')
DRN_cell_12_char = pd.read_csv('DRN-serotonin_Female_spikes\CELL 12 DEEP 4850_Spikes.csv')
DRN_cell_13_char = pd.read_csv('DRN-serotonin_Female_spikes\CELL 13 DEEP 4983_Spikes.csv')
DRN_cell_14_char = pd.read_csv('DRN-serotonin_Female_spikes\CELL 14 DEEP 6318_Spikes.csv')
DRN_cell_15_char = pd.read_csv('DRN-serotonin_Female_spikes\CELL 15 DEEP 6426_Spikes.csv')
DRN_cell_16_char = pd.read_csv('DRN-serotonin_Female_spikes\CELL 16 DEEP 6658_Spikes.csv')
```

```python
import matplotlib.pyplot as plt

#cell_11_char_transposed = cell_11_char.T

# Построение графика
plt.figure(figsize=(10, 6))
for spike_num, values in DRN_cell_11_char.iterrows():
    plt.plot(values, label=f"Spike {spike_num}")

plt.title("Spike Characteristics Over Time")
plt.xlabel("Time (seconds)")
plt.ylabel("Spike Value")
plt.grid(True)
plt.show()
```

```python
def graphs(cell):
    plt.figure(figsize=(10, 6))
    for spike_num, values in cell.iterrows():
        plt.plot(values, label=f"Spike {spike_num}")
    plt.title("Spike Characteristics Over Time")
    plt.xlabel("Time (seconds)")
    plt.ylabel("Spike Value")
    plt.grid(True)
    plt.show()
```

```python
def graphs_t(cell):
    cell_1 = cell.T
    plt.figure(figsize=(10, 6))
    for spike_num, values in cell_1.iterrows():
        plt.plot(values, label=f"Spike {spike_num}")
    plt.title("Spike Characteristics Over Time")
    plt.xlabel("Time (seconds)")
    plt.ylabel("Spike Value")
    plt.grid(True)
    plt.show()
```

```python
result_DRN = pd.concat([DRN_cell_11_char, DRN_cell_12_char, DRN_cell_13_char, DRN_cell_14_char, DRN_cell_15_char, DRN_cell_16_char], axis=1) 
```

```python
result_DRN
```

```python
from sklearn.utils import shuffle
result_DRN_shuffeled = shuffle(result_DRN)
```

```python
graphs(result_DRN_shuffeled)
```

```python
result_DRN_shuffeled
```

```python
fft_results = pd.DataFrame()

for spike_num, values in result_DRN_shuffeled.iterrows():
    # Выполнение преобразования Фурье
    spectrum = np.fft.fft(values)
    # Сохранение амплитудного спектра
    fft_results[f"Spike {spike_num}"] = np.abs(spectrum)

# Транспонирование для удобства визуализации
fft_results = fft_results.T
```

```python

plt.figure(figsize=(12, 6))
for spike_num, values in fft_results.iterrows():
    plt.plot(values, label=spike_num)

plt.title("Amplitude Spectrum after FFT")
plt.xlabel("Frequency Bin")
plt.ylabel("Amplitude")

plt.grid(True)
plt.show()
```
